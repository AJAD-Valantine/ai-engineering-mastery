```python
# Imports

from typing import TypedDict

from langgraph.graph import END, START, StateGraph
```

```python
# Define State

class ReportState(TypedDict):
    file_path: str
    word_count: int
    line_count: int
    char_count: int
    report: str
```

```python
# Node 01 - Word Count

def count_words(state: ReportState) -> ReportState:
    with open(state["file_path"], "r", encoding="utf-8") as f:
        text: str = f.read()
    state["word_count"] = len(text.split())
    return state
```

```python
# Node 02 - Character Count

def count_chars(state: ReportState) -> ReportState:
    with open(state["file_path"], "r", encoding="utf-8") as f:
        text: str = f.read()
    state["char_count"] = len(text)
    return state
```

```python
# Node 03 - Line Count

def count_lines(state: ReportState) -> ReportState:
    with open(state["file_path"], "r", encoding="utf-8") as f:
        text: str = f.read()
    state["line_count"] = len(text.splitlines())
    return state
```

```python
# Node 04 - Generate Report

def generate_report(state: ReportState) -> ReportState:
    state["report"] = f"File: {state['file_path']}\nWords: {state['word_count']}\nLines: {state['line_count']}\nCharacters: {state['char_count']}"
    return state
```

```python
# Define Graph

graph = StateGraph(ReportState)
```

```python
# Add Nodes

graph.add_node("count_words", count_words)
graph.add_node("count_chars", count_chars)
graph.add_node("count_lines", count_lines)
graph.add_node("generate_report", generate_report)
```

```python
# Add Edges

graph.add_edge(START, "count_words")
graph.add_edge("count_words", "count_chars")
graph.add_edge("count_chars", "count_lines")
graph.add_edge("count_lines", "generate_report")
graph.add_edge("generate_report", END)
```

```python
# Compile the Graph

workflow = graph.compile()
```

```python
# Execute Graph

initial_state = {"file_path": "example.txt"}
final_state = workflow.invoke(initial_state) # pyright: ignore[reportArgumentType]
print(final_state)
```
